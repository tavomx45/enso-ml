Zona costera ecuatoriana

```python
coast = ds.sel(
    latitude=slice(1.5, -5.0),
    longitude=slice(-81.5, -79.0)
)
```

```python
tp = coast["tp"]

valid_cells = tp.notnull().sum(dim=["latitude", "longitude"])

valid_cells
```

```python
print("Celdas válidas mínimas:", int(valid_cells.min()))
print("Celdas válidas máximas:", int(valid_cells.max()))
```

Comprobacion si es valor mensual acumulado o promediado

```python
tp_mean = tp.mean(
    dim=["latitude", "longitude"],
    skipna=True
)
```

```python
float(tp_mean.sel(valid_time="1998-01-01"))
```

```python
tp_mean.attrs
```

```python
float(tp_mean.sel(valid_time="1998-01-01"))
```

```python
float(tp_mean.sel(valid_time="2013-01-01"))
```

convertir la precipitacion en metros por dia a precipitacion mensual total

```python
days_in_month = tp_mean["valid_time"].dt.days_in_month

tp_monthly_mm = (
    tp_mean
    * 1000
    * days_in_month
)
```

```python
tp_monthly_mm.name = "precipitation_mm"

tp_monthly_mm.attrs = {
    "long_name": "Monthly total precipitation - Ecuadorian coast",
    "units": "mm/month"
}
```

```python
print(
    "1998-01:",
    float(tp_monthly_mm.sel(valid_time="1998-01-01"))
)

print(
    "2013-01:",
    float(tp_monthly_mm.sel(valid_time="2013-01-01"))
)
```

Contruccion de nuestro dataframe

```python
precip_df = (
    tp_monthly_mm
    .to_dataframe()
    .reset_index()
    .rename(columns={"valid_time": "date"})
)
```

Grafico de la serie temporal

```python
fig, ax = plt.subplots(figsize=(14, 5))

ax.plot(
    precip_df["date"],
    precip_df["precipitation_mm"],
    linewidth=0.8
)

ax.set_title(
    "Precipitación mensual promedio - Costa ecuatoriana"
)
ax.set_xlabel("Fecha")
ax.set_ylabel("Precipitación (mm/mes)")

plt.show()
```

Climatologia mensual

```python
precip_df["month"] = precip_df["date"].dt.month

monthly_climatology = (
    precip_df
    .groupby("month")["precipitation_mm"]
    .mean()
)
```

```python
fig, ax = plt.subplots(figsize=(10, 5))

ax.bar(
    monthly_climatology.index,
    monthly_climatology.values
)

ax.set_title(
    "Climatología mensual de precipitación - Costa ecuatoriana"
)
ax.set_xlabel("Mes")
ax.set_ylabel("Precipitación media (mm/mes)")
ax.set_xticks(range(1, 13))

plt.show()
```

Inspeccion de los eventos ENSO conocidos

```python
def plot_precipitation_period(df, start, end):
    subset = df[
        df["date"].between(start, end)
    ]

    fig, ax = plt.subplots(figsize=(12, 5))

    ax.plot(
        subset["date"],
        subset["precipitation_mm"],
        marker="o",
        markersize=3
    )

    ax.set_title(
        f"Precipitación: {start} → {end}"
    )
    ax.set_xlabel("Fecha")
    ax.set_ylabel("Precipitación (mm/mes)")

    plt.show()
```

```python
plot_precipitation_period(
    precip_df,
    "1981-01-01",
    "1984-12-01"
)
```

```python
plot_precipitation_period(
    precip_df,
    "1996-01-01",
    "1999-12-01"
)
```

```python
plot_precipitation_period(
    precip_df,
    "2014-01-01",
    "2017-12-01"
)
```
