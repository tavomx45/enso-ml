## Transformacion de los datos

```python
for name, df in {
    "nino12": nino12,
    "nino3": nino3,
    "nino34": nino34,
    "nino4": nino4,
    "soi": soi,
    "tni": tni,
}.items():
    print(name, "->", df.columns.tolist())
```

Transformar a serie temporal y variables sentinelas

```python
def prepare_csv_series(df, variable_name):
    """
    Homogeneiza una serie temporal de dos columnas:
    fecha y valor.
    """
    df = df.copy()

    # Homogeneizar nombres de columnas
    df.columns = ["date", variable_name]

    # Convertir fecha de string a datetime
    df["date"] = pd.to_datetime(df["date"])

    # Convertir sentinelas NOAA en valores faltantes
    df.loc[df[variable_name] <= -90, variable_name] = np.nan

    return df
```

```python
# aplicacion de la funcion
nino12_clean = prepare_csv_series(nino12, "nino12")
nino3_clean = prepare_csv_series(nino3, "nino3")
nino34_clean = prepare_csv_series(nino34, "nino34")
nino4_clean = prepare_csv_series(nino4, "nino4")
soi_clean = prepare_csv_series(soi, "soi")
tni_clean = prepare_csv_series(tni, "tni")
```

```python
nino34_clean.head()
```

```python
nino34_clean.dtypes
```

```python
# comprobar si los sentinelas se hiceron nan
for name, df in {
    "nino12": nino12_clean,
    "nino3": nino3_clean,
    "nino34": nino34_clean,
    "nino4": nino4_clean,
    "soi": soi_clean,
    "tni": tni_clean,
}.items():
    print(
        f"{name:8s} | "
        f"filas: {len(df):4d} | "
        f"missing: {df[name].isna().sum():3d} | "
        f"{df['date'].min().date()} -> {df['date'].max().date()}"
    )
```

Transformar los datos de MEI.v2

```python
mei_wide = pd.read_csv(
    DATA_RAW / "meiv2.data",
    sep=r"\s+",
    skiprows=1,
    header=None
)
```

```python
mei_wide.head()
```

```python
mei_wide.shape
```

poniendo nombres de columnas

```python
month_names = [
    "jan", "feb", "mar", "apr", "may", "jun",
    "jul", "aug", "sep", "oct", "nov", "dec"
]

mei_wide.columns = ["year"] + month_names
```

```python
mei_wide.head()
```

transformar el formato del tiempo

```python
mei_long = mei_wide.melt(
    id_vars="year",
    var_name="month",
    value_name="mei"
)
```

```python
mei_long.head(15)
```

Construccion de la fecha

```python
# meses a numeros
mont_map = {
    "jan": 1,
    "feb": 2,
    "mar": 3,
    "apr": 4,
    "may": 5,
    "jun": 6,
    "jul": 7,
    "aug": 8,
    "sep": 9,
    "oct": 10,
    "nov": 11,
    "dec": 12
}

mei_long["month_num"] = mei_long["month"].map(mont_map)
```

```python
mei_long["date"] = pd.to_datetime(
    dict(
        year=mei_long["year"],
        month=mei_long["month_num"],
        day=1
    )
)
```

```python
mei_clean = (
    mei_long[["date", "mei"]]
    .sort_values("date")
    .reset_index(drop=True)
)
```

```python
mei_clean.head()
```

sentinelas de mei

```python
mei_clean["mei"].min()
```

```python
mei_clean.loc[mei_clean["mei"] <= -90, "mei"] = np.nan
```

Se tienen todos los datos homogeneos

```python
clean_datasets = {
    "nino12": nino12_clean,
    "nino3": nino3_clean,
    "nino34": nino34_clean,
    "nino4": nino4_clean,
    "soi": soi_clean,
    "tni": tni_clean,
    "mei": mei_clean,
    "oni": oni_clean,
}

summary = []

for name, df in clean_datasets.items():
    summary.append({
        "variable": name,
        "observations": len(df),
        "start": df["date"].min(),
        "end": df["date"].max(),
        "missing": df[name].isna().sum()
    })

summary_df = pd.DataFrame(summary)

summary_df
```

```python
# rangos temporales efectivos
for name, df in clean_datasets.items():
    print(
        f"{name:8s}: "
        f"{df['date'].min().date()} → "
        f"{df['date'].max().date()} | "
        f"NaN = {df[name].isna().sum()}"
    )
```

### Construir el dataset maestro

```python
from functools import reduce

dataframes = [
    nino12_clean,
    nino3_clean,
    nino34_clean,
    nino4_clean,
    soi_clean,
    tni_clean,
    mei_clean,
    oni_clean,
]

enso_master = reduce(
    lambda left, right: pd.merge(
        left,
        right,
        on="date",
        how="outer"
    ),
    dataframes
)

enso_master = (
    enso_master
    .sort_values("date")
    .reset_index(drop=True)
)
```

```python
enso_master.tail()
```

```python
enso_master["date"].duplicated().sum()
```

```python
for name, df in clean_datasets.items():
    duplicates = df["date"].duplicated().sum()
    print(f"{name:8s}: {duplicates} fechas duplicadas")
```

```python
expected_dates = pd.date_range(
    start=enso_master["date"].min(),
    end=enso_master["date"].max(),
    freq="MS"
)

len(expected_dates)
```

```python
missing_dates = expected_dates.difference(enso_master["date"])

len(missing_dates)
```

### Construccion de los dos datasets

Dataset historico (mayor cantidad de datos)

```python
historical_columns = [
    "date",
    "nino12",
    "nino3",
    "nino34",
    "nino4",
    "soi",
    "tni",
]

enso_historical = enso_master[historical_columns].copy()
```

```python
enso_historical.head()
```

```python
enso_historical = (
    enso_historical
    .dropna()
    .reset_index(drop=True)
)
```

```python
print(enso_historical.shape)
print(
    enso_historical["date"].min(),
    "→",
    enso_historical["date"].max()
)
```

Dataset moderno

```python
modern_columns = [
    "date",
    "nino12",
    "nino3",
    "nino34",
    "nino4",
    "soi",
    "tni",
    "mei",
    "oni",
]

enso_modern = (
    enso_master.loc[
        enso_master["date"] >= "1979-01-01",
        modern_columns
    ]
    .dropna()
    .reset_index(drop=True)
)

print(enso_modern.shape)
print(
    enso_modern["date"].min(),
    "→",
    enso_modern["date"].max()
)
```

Validacion final

```python
dataset_comparison = pd.DataFrame({
    "dataset": [
        "Master",
        "Historical",
        "Modern"
    ],
    "observations": [
        len(enso_master),
        len(enso_historical),
        len(enso_modern)
    ],
    "start": [
        enso_master["date"].min(),
        enso_historical["date"].min(),
        enso_modern["date"].min()
    ],
    "end": [
        enso_master["date"].max(),
        enso_historical["date"].max(),
        enso_modern["date"].max()
    ],
    "variables": [
        len(enso_master.columns) - 1,
        len(enso_historical.columns) - 1,
        len(enso_modern.columns) - 1
    ]
})

dataset_comparison
```

guardar los datos procesados

```python
DATA_PROCESSED = PROJECT_ROOT / "data" / "processed"

enso_master.to_csv(
    DATA_PROCESSED / "enso_master.csv",
    index=False
)

enso_historical.to_csv(
    DATA_PROCESSED / "enso_historical.csv",
    index=False
)

enso_modern.to_csv(
    DATA_PROCESSED / "enso_modern.csv",
    index=False
)
```


## Conclusiones del entendimiento de datos

- Las fuentes contienen información climática mensual relacionada con ENSO.
- Los índices Niño 1+2, Niño 3, Niño 3.4, Niño 4, SOI y TNI poseen
  series históricas extensas.
- ONI comienza en 1950 y MEI.v2 en 1979.
- Las fuentes originales utilizan diferentes estructuras de almacenamiento.
- Los códigos sentinela negativos fueron convertidos a valores faltantes.
- Las fechas fueron homogeneizadas a frecuencia mensual.
- Se construyó un dataset maestro mediante integración temporal.
- Se conservaron dos conjuntos analíticos:
  - **Historical:** mayor cobertura temporal y menor número de indicadores.
  - **Modern:** menor cobertura temporal pero incorpora MEI.v2 y ONI.
- Estos conjuntos todavía no representan matrices finales de Machine Learning.
